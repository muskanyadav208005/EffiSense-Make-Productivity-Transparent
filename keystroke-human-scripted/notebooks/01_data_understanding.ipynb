```python
import os
import glob
from collections import Counter

folder = r"C:\Users\DELL\OneDrive\Desktop\keystroke-human-scripted\data\raw\between-subject"

arff_files = glob.glob(os.path.join(folder, "**", "*.arff"), recursive=True)

total_files = len(arff_files)
total_rows = 0
label_counts = Counter()
usable_files = 0

for file in arff_files:
    try:
        with open(file, "r", encoding="utf-8", errors="ignore") as f:
            lines = f.readlines()

        # Find the @DATA section
        data_start = None

        for i, line in enumerate(lines):
            if line.strip().upper() == "@DATA":
                data_start = i + 1
                break

        if data_start is None:
            continue

        file_labels = Counter()

        for line in lines[data_start:]:
            line = line.strip()

            if not line or line.startswith("%"):
                continue

            # Last value is the RESULT label
            label = line.split(",")[-1].strip()

            if label in ["legitimate", "impostor"]:
                file_labels[label] += 1
                label_counts[label] += 1
                total_rows += 1

        if sum(file_labels.values()) > 0:
            usable_files += 1

    except Exception as e:
        print("Error:", file)

print("================================")
print("Total ARFF files:", total_files)
print("Usable ARFF files:", usable_files)
print("Total samples:", total_rows)
print("================================")
print("Label distribution:")
print(label_counts)
print("================================")
```

```python
import os
import glob
from collections import Counter

folder = r"C:\Users\DELL\OneDrive\Desktop\keystroke-human-scripted\data\raw\between-subject"

arff_files = glob.glob(
    os.path.join(folder, "**", "*.arff"),
    recursive=True
)

result_attributes = Counter()
file_patterns = Counter()

for file in arff_files:
    try:
        with open(file, "r", encoding="utf-8", errors="ignore") as f:
            for line in f:
                line = line.strip()

                if line.upper().startswith("@ATTRIBUTE"):
                    result_attributes[line] += 1

                if line.upper() == "@DATA":
                    break

        name = os.path.basename(file)

        # Count important filename patterns
        parts = name.split("-")

        if "EVALUATE" in name:
            file_patterns["EVALUATE"] += 1
        if "TRAIN" in name.upper():
            file_patterns["TRAIN"] += 1
        if "MIXED" in name:
            file_patterns["MIXED"] += 1

    except:
        pass

print("TOTAL ARFF FILES:", len(arff_files))

print("\nRESULT/ATTRIBUTE DEFINITIONS:")
for x, count in result_attributes.items():
    if "RESULT" in x.upper():
        print(count, "->", x)

print("\nFILE PATTERNS:")
print(file_patterns)
```

```python
import os
import glob
from collections import Counter

folder = r"C:\Users\DELL\OneDrive\Desktop\keystroke-human-scripted\data\raw\within-subject"

arff_files = glob.glob(
    os.path.join(folder, "**", "*.arff"),
    recursive=True
)

print("Total ARFF files:", len(arff_files))

total_labels = Counter()
files_with_both = 0
files_with_impostor = 0

for file in arff_files:

    try:
        with open(file, "r", encoding="utf-8", errors="ignore") as f:
            found_data = False
            labels = []

            for line in f:
                line = line.strip()

                if line.upper() == "@DATA":
                    found_data = True
                    continue

                if found_data and line and not line.startswith("%"):
                    label = line.split(",")[-1].strip()
                    labels.append(label)

            counts = Counter(labels)

            total_labels.update(counts)

            if "impostor" in counts:
                files_with_impostor += 1

            if "impostor" in counts and "legitimate" in counts:
                files_with_both += 1

    except Exception as e:
        pass

print("\nACTUAL DATA LABEL DISTRIBUTION:")
print(total_labels)

print("\nFiles containing impostor rows:", files_with_impostor)
print("Files containing BOTH classes:", files_with_both)
```

```python
import os
import glob
import pandas as pd
import zipfile

base = r"C:\Users\DELL\OneDrive\Desktop\keystroke-human-scripted\data\raw"
within = os.path.join(base, "within-subject")
zip_file = os.path.join(base, "within-subject.zip")

# --------------------------------------------------
# 1. Check ARFF filenames for TRAIN / TRAINING
# --------------------------------------------------

arff_files = glob.glob(
    os.path.join(within, "**", "*.arff"),
    recursive=True
)

train_files = [
    f for f in arff_files
    if "TRAIN" in os.path.basename(f).upper()
]

evaluate_files = [
    f for f in arff_files
    if "EVALUATE" in os.path.basename(f).upper()
]

print("Total ARFF files:", len(arff_files))
print("TRAIN/TRAINING ARFF files:", len(train_files))
print("EVALUATE ARFF files:", len(evaluate_files))

print("\nExample TRAIN files:")
for f in train_files[:5]:
    print(os.path.basename(f))


# --------------------------------------------------
# 2. Check ZIP contents directly
# --------------------------------------------------

print("\nChecking ZIP contents...")

with zipfile.ZipFile(zip_file, "r") as z:

    names = z.namelist()

    train_zip = [
        n for n in names
        if n.lower().endswith(".arff")
        and "train" in os.path.basename(n).lower()
    ]

    evaluate_zip = [
        n for n in names
        if n.lower().endswith(".arff")
        and "evaluate" in os.path.basename(n).lower()
    ]

    print("ARFF files inside ZIP:", 
          sum(n.lower().endswith(".arff") for n in names))

    print("TRAIN ARFF files inside ZIP:", len(train_zip))
    print("EVALUATE ARFF files inside ZIP:", len(evaluate_zip))

    print("\nExample TRAIN files inside ZIP:")
    for n in train_zip[:5]:
        print(n)


# --------------------------------------------------
# 3. Check ResultsBySentence CSVs
# --------------------------------------------------

print("\nChecking ResultsBySentence CSV files...")

csv_files = glob.glob(
    os.path.join(within, "**", "*ResultsBySentence.csv"),
    recursive=True
)

print("ResultsBySentence files:", len(csv_files))

for file in csv_files:

    df = pd.read_csv(file)

    print("\n", os.path.basename(file))
    print("Shape:", df.shape)

    if "class" in df.columns:
        print("Classes:")
        print(df["class"].value_counts())

    if "classified_as_legitimate" in df.columns:
        print("classified_as_legitimate:")
        print(df["classified_as_legitimate"].value_counts())
```

```python
import os
from collections import Counter

folder = r"C:\Users\DELL\OneDrive\Desktop\keystroke-human-scripted\data\raw\within-subject"

file_types = Counter()
examples = {}

for root, dirs, files in os.walk(folder):
    for file in files:
        ext = os.path.splitext(file)[1].lower()
        file_types[ext] += 1

        if ext not in examples:
            examples[ext] = []

        if len(examples[ext]) < 5:
            examples[ext].append(os.path.join(root, file))

print("FILE TYPE COUNTS:")
for ext, count in file_types.items():
    print(f"{ext or '[no extension]'} : {count}")

print("\nEXAMPLES:")
for ext, files in examples.items():
    print(f"\n{ext or '[no extension]'}")
    for f in files:
        print(" ", f)


# ---------------------------------------------
# Check ALL ResultsBySentence files
# ---------------------------------------------

print("\n\nRESULTS BY SENTENCE:")

for root, dirs, files in os.walk(folder):

    for file in files:

        if file.endswith("ResultsBySentence.csv"):

            path = os.path.join(root, file)

            import pandas as pd

            df = pd.read_csv(path)

            print("\n", file)
            print("Rows:", len(df))
            print("Class:", df["class"].value_counts().to_dict())

            print("Columns:")
            print(list(df.columns))
```
