```python
import glob

arff_files = glob.glob(os.path.join(between_actual, "*.arff"))

print("Number of ARFF files:", len(arff_files))


def read_arff(file_path):
    rows = []
    data_started = False

    with open(file_path, "r") as f:
        for line in f:
            line = line.strip()

            if not line:
                continue

            if line.upper() == "@DATA":
                data_started = True
                continue

            if data_started and not line.startswith("%"):
                rows.append(line.split(","))

    return rows


# Read the first file to determine columns
first_rows = read_arff(arff_files[0])

print("Rows in first file:", len(first_rows))
print("Columns in first row:", len(first_rows[0]))
print("Last column:", first_rows[0][-1])
```

```python
all_rows = []

for i, file_path in enumerate(arff_files):

    rows = read_arff(file_path)
    all_rows.extend(rows)

    if (i + 1) % 1000 == 0:
        print(f"Processed {i + 1} / {len(arff_files)} files")

print("\nTotal rows:", len(all_rows))
print("Columns:", len(all_rows[0]))
```

```python
feature_names = [
    "FT_DM", "FT_DE", "FT_DMin", "FT_DC", "FT_Z",
    "FT_HM", "FT_HE", "FT_HMin", "FT_HC", "FT_HZ",
    "FT_DIE", "FT_DIL", "FT_RAll", "FT_RMix", "FT_R1",
    "HT_DM", "HT_DE", "HT_DMin", "HT_DC", "HT_Z",
    "HT_HM", "HT_HE", "HT_HMin", "HT_HC", "HT_HZ",
    "HT_DIE", "HT_DIL", "HT_RAll", "HT_RMix", "HT_R1",
    "RESULT"
]

df = pd.DataFrame(all_rows, columns=feature_names)

# Convert numeric columns
numeric_features = feature_names[:-1]

df[numeric_features] = df[numeric_features].apply(
    pd.to_numeric,
    errors="coerce"
)

print("Shape:", df.shape)

print("\nLabels:")
print(df["RESULT"].value_counts())

print("\nMissing values:")
print(df.isnull().sum().sum())
```

```python
impostor_dfs = []

for file in results_files:
    temp = pd.read_csv(file)
    temp = temp[["user", "sentenceID", "length", "hold_times", "flight_times"]].copy()
    impostor_dfs.append(temp)

impostor = pd.concat(impostor_dfs, ignore_index=True)

print("Total impostor samples:", len(impostor))
print(impostor.head())
```

```python
def parse_times(value):
    values = str(value).split("-")
    
    # Remove X from flight times
    values = [v for v in values if v != "X" and v != ""]
    
    return np.array([float(v) for v in values], dtype=float)


def extract_features(row):
    hold = parse_times(row["hold_times"])
    flight = parse_times(row["flight_times"])
    
    features = {
        "hold_mean": np.mean(hold),
        "hold_std": np.std(hold),
        "hold_min": np.min(hold),
        "hold_max": np.max(hold),

        "flight_mean": np.mean(flight),
        "flight_std": np.std(flight),
        "flight_min": np.min(flight),
        "flight_max": np.max(flight),

        "hold_median": np.median(hold),
        "flight_median": np.median(flight),

        "hold_range": np.max(hold) - np.min(hold),
        "flight_range": np.max(flight) - np.min(flight),

        "num_keystrokes": len(hold),
    }
    
    return pd.Series(features)


impostor_features = impostor.apply(extract_features, axis=1)

print("Shape:", impostor_features.shape)
print("\nFirst 5 rows:")
print(impostor_features.head())
```

```python
# Combine all ResultsBySentence files
all_results = []

for file in results_files:
    temp = pd.read_csv(file)

    temp = temp[
        ["user", "sentenceID", "length", "class",
         "classified_as_legitimate", "hold_times", "flight_times"]
    ].copy()

    all_results.append(temp)

results = pd.concat(all_results, ignore_index=True)

print("Shape:", results.shape)
print("\nClass:")
print(results["class"].value_counts())

print("\nClassified as legitimate:")
print(results["classified_as_legitimate"].value_counts())
```

```python
# Extract raw timing features from all evaluation samples

def parse_times(value):
    values = str(value).split("-")
    values = [v for v in values if v not in ["X", ""]]
    return np.array([float(v) for v in values], dtype=float)

def extract_features(row):
    hold = parse_times(row["hold_times"])
    flight = parse_times(row["flight_times"])

    return pd.Series({
        "hold_mean": np.mean(hold),
        "hold_std": np.std(hold),
        "hold_min": np.min(hold),
        "hold_max": np.max(hold),
        "hold_median": np.median(hold),
        "hold_range": np.ptp(hold),

        "flight_mean": np.mean(flight),
        "flight_std": np.std(flight),
        "flight_min": np.min(flight),
        "flight_max": np.max(flight),
        "flight_median": np.median(flight),
        "flight_range": np.ptp(flight),

        "num_keystrokes": len(hold)
    })

X_raw = results.apply(extract_features, axis=1)

print("Feature shape:", X_raw.shape)
display(X_raw.head())
```

```python
# Check for invalid values
print("Missing values:")
print(X_raw.isnull().sum().sum())

print("\nInfinite values:")
print(np.isinf(X_raw).sum().sum())
```
