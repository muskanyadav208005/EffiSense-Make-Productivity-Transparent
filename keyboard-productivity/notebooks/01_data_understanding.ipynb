```python
total_available = window_counts["windows"].sum()

target = 10000
selection_ratio = target / total_available

print("Available windows:", total_available)
print("Target windows:", target)
print("Selection ratio:", selection_ratio)

window_counts["target_float"] = (
    window_counts["windows"] * selection_ratio
)

print("\nTarget windows per group:")
print(
    window_counts[
        ["participant", "session", "windows", "target_float"]
    ].head(10).to_string(index=False)
)
```

```python
import numpy as np

window_counts["target_windows"] = np.floor(
    window_counts["target_float"]
).astype(int)

print("Total after flooring:",
      window_counts["target_windows"].sum())

print("\nWindows still needed:",
      10000 - window_counts["target_windows"].sum())
```

```python
# Fractional part of each group's ideal allocation
window_counts["fraction"] = (
    window_counts["target_float"]
    - window_counts["target_windows"]
)

# Give one extra window to the 87 groups
# with the largest fractional parts
remaining = 10000 - window_counts["target_windows"].sum()

top_groups = window_counts["fraction"].nlargest(remaining).index

window_counts.loc[top_groups, "target_windows"] += 1

print("Final allocated windows:",
      window_counts["target_windows"].sum())

print("\nAllocation statistics:")
print(window_counts["target_windows"].describe())

print("\nGroups receiving an extra window:",
      (window_counts["fraction"].rank(method="first", ascending=False) <= remaining).sum())
```

```python
windows = []

for (participant, session), group in clean_df.groupby(
    ["participant", "session"], sort=False
):
    group = group.reset_index(drop=True)

    # Number of windows allocated to this participant-session
    n_windows = int(
        window_counts.loc[
            (window_counts["participant"] == participant) &
            (window_counts["session"] == session),
            "target_windows"
        ].iloc[0]
    )

    # Create consecutive non-overlapping 50-event windows
    for window_id in range(n_windows):
        start = window_id * 50
        end = start + 50

        window = group.iloc[start:end].copy()

        # Safety check
        if len(window) == 50:
            windows.append({
                "participant": participant,
                "session": session,
                "window_id": window_id,
                "data": window
            })

print("Total windows created:", len(windows))
```
